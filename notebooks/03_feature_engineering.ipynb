```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\HP\Desktop\RealEstatePricePrediction\data\processed\cleaned_data.csv")
```

```python
print(df.shape)
df.head()
```

```python
print("Total unique locations:", df["location"].nunique())
print("\nLocation counts:\n", df["location"].value_counts().tail(20))
```

```python
location_counts = df["location"].value_counts()
rare_locations = location_counts[location_counts <= 10].index
df["location"] = df["location"].apply(lambda x: "other" if x in rare_locations else x)

print("Unique locations after grouping:", df["location"].nunique())
```

- 752 locations before grouping, many with just 1 house in that particular location
- 176 locations after grouping, all with atleast 11 houses

```python
#one hot encoding for location
df = pd.get_dummies(df, columns=["location"])
print("Shape after encoding:", df.shape)
print(df.head())
```

```python
df = df.astype({col: int for col in df.select_dtypes(include='bool').columns})
print(df.head())
```
